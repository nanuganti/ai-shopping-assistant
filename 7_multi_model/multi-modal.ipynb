```python
from dotenv import load_dotenv
from langchain_groq import ChatGroq
from langchain_core.messages import HumanMessage
import os
import base64

load_dotenv()
```

```python
# Encode image and send it to vision model 
# read and convert to base64 encoding
with open("blood_work.png", "rb") as f:
    image_data = f.read()
    image_b64 = base64.b64encode(image_data).decode("utf-8")

## print 200 chars
image_b64[:200]
```

```python
from langchain_openai import ChatOpenAI
from langchain_core.messages import HumanMessage
import os

# 1. Clean the base64 string to prevent payload corruption
clean_b64 = image_b64.replace("\n", "").replace("\r", "").strip()

# 2. Grab your key from your environment
# Note: You can also pass your raw string "sk-or-..." directly here if it isn't saved in your environment yet
my_openrouter_key = os.environ.get("OPENROUTER_API_KEY") 

# 3. Explicitly target OpenRouter while feeding the required parameters
llm = ChatOpenAI(
    openai_api_base="https://openrouter.ai",
    openai_api_key=my_openrouter_key,
    api_key=my_openrouter_key,        # Explicitly pass to both arguments to bypass strict validation
    model="openai/gpt-oss-120b"
)

# 4. Construct the multimodal payload
message = HumanMessage(
    content=[
        {
            "type": "image_url",
            "image_url": {
                "url": f"data:image/png;base64,{clean_b64}"
            }
        },
        {
            "type": "text",
            "text": "This is a blood work report. Extract all test results and flag any abnormalities?"
        }
    ]
)

# 5. Fire the request
response = llm.invoke([message])
print(response.content)
```
