```python
# read file content and print 200 chars
with open("blood_work.txt", "r") as f:
    blood_work_content = f.read()
    print(blood_work_content[:200])
```

```python
from langchain_groq import ChatGroq

llm = ChatGroq(model="qwen/qwen3.6-27b")

extraction_prompt = f"""
You are a helpful assistant that provides information about blood work analysis, look for Normal, Low and High values.
Here is the blood work content:

format your response in the following format:
- Test Name: value, Result: HIGH/LOW/NORMAL, Reference: range

blood work content:
{blood_work_content}
"""

extraction_response = llm.invoke(extraction_prompt)
print("Extraction Response from LLM:")
print(extraction_response.text)
```

```python
diet_prompt = f"""
You are a dietitian assistant that provides dietary recommendations based on blood work analysis.

Review the following blood work:
{blood_work_content}

Provide practical dietary recommendations based on abnormal results.
Include:
- Foods to eat more often
- Foods to limit

Do not diagnose or prescribe treatment.
"""

diet_response = llm.invoke(diet_prompt)

print("Dietary Recommendations:")
print(diet_response.text)
```
