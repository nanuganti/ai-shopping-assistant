```python
import chromadb

client = chromadb.Client()
```

```python

collection = client.create_collection("news_articles")
```

```python
collection.add(
    ids=["1", "2", "3", "4", "5"],
    documents=[
        "The stock market saw a significant increase today, with major indices closing at record highs.",
        "A new study reveals the benefits of a plant-based diet for long-term health.",
        "The local football team won their championship game in a thrilling overtime victory.",
        "Scientists have discovered a new species of deep-sea fish in the Pacific Ocean.",
        "The city council has approved a new public transportation initiative to reduce traffic congestion."
        "spaceX has successfully launched its latest rocket, marking a milestone in commercial space travel."
    ],
    metadatas=[
        {"source": "Financial Times", "date": "2023-10-01"},
        {"source": "Health Journal", "date": "2023-10-02"},
        {"source": "Sports Daily", "date": "2023-10-03"},
        {"source": "Science News", "date": "2023-10-04"},
        {"source": "City Council Report", "date": "2023-10-05"}
    ]);
```

```python
## fetch documents and embeddings from the collection
results = collection.get(
    ids=["1", "2", "3", "4", "5"],
    include=["documents", "metadatas", "embeddings"]
)

## print Embeddings

for i, emb in enumerate(results['embeddings']):
    print(f"Document ID: {results['ids'][i]}")
    print(f"Text: {results['documents'][i]}")
    print("Embedding Length:", len(emb))
    print("First 10 elements of Embedding:", emb[:10])
```

```python
## query collection 

results = collection.query(
    query_texts=["What are the benefits of a plant-based diet?"],)
```
